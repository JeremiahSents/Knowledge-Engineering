# Part 3: Knowledge Engineering — Smart Traffic Light Navigation Agent

*Adapted from [Zainabu Muti](https://www.linkedin.com/in/zainabu-muti-3789591b3/)*

**Step 1: Identify the Task**

Decide what a car should do based on the traffic light colour.

* **Input:** observed light colour
* **Output:** driving action
* **Goal:** safe decisions with minimal risk

**Step 2: Assemble Relevant Knowledge**

* The colour of the traffic light
* The correct action for each colour, and whether it is safe to go

**Source:** standard road traffic rules. No APIs or web scraping were used.

**Step 3: Define the Vocabulary**

* **Predicates:** `Light(color)`, `Action(color, action)`, `CanGo(color)`
* **Constants (colours):** Red, Yellow, Green
* **Constants (actions):** Stop, SlowDown, Go

**Step 4: Encode General Knowledge about the Domain** `[3 marks]`

`rules` encodes `Action(color, action)`; `CanGo()` encodes `CanGo(color)`.

```python
# Action(color, action)
rules = {
    "Red": "Stop — it’s unsafe to move.",
    "Yellow": "Slow down — prepare to stop.",
    "Green": "Go — it’s safe to proceed.",
}

# CanGo(color): True only when it is safe to proceed
can_go = {"Red": False, "Yellow": False, "Green": True}

def CanGo(color):
    return can_go.get(color, False)  # unknown signal -> not safe

for color in rules:
    print(f"{color:6} CanGo={CanGo(color)}  ➜ {rules[color]}")
```

**Step 5: Add New Facts to the Knowledge Base** `[4 marks]`

```python
def show_kb():
    print("\n--- Current Knowledge Base ---")
    for light, action in rules.items():
        print(f"{light:10} ➜ {action}")
    print("-------------------------------")

def add_new_fact(color, action):
    """Add a new fact directly to the KB (- without user input)."""
    rules[color.capitalize()] = action
    print(f"Added fact: '{color.capitalize()}' ➜ '{action}'")

# Example of adding facts
add_new_fact("Flashing", "Stop and check intersection")
add_new_fact("Blinking", "Proceed slowly with caution")

def add_fact():
    """Add new knowledge (facts) to the KB."""
    new_light = input("Enter new traffic light color: ").capitalize()
    new_action = input(f"What should the agent do when light is {new_light}? ")
    rules[new_light] = new_action
    print(f" New fact added: '{new_light}' -> '{new_action}'")

add_fact()

# View updated KB
show_kb()
```

**Steps 6 & 7: Encode Specific Problem Instance (Facts) & Pose Queries and Get Answers** `[3 marks]`



```python
# Step 6: The current observation from sensors/camera
current_light = input("Enter the current traffic light color: ").capitalize()

# Step 7: The agent queries the knowledge base (rules dictionary)
def traffic_advice(light):
    action = rules.get(light, "Unknown signal — proceed with caution or stop.")
    return action

# Output the result
print("\n=== Traffic Light Decision System ===")
print(f"Observed Light: {current_light}")
print("Agent Decision:", traffic_advice(current_light))
```

**Step 8: Querying the knowledge base**

```python
#ask questions to get actions based on known facts (in this case; lights)
print("Agent Decision for 'Flashing':", traffic_advice("Flashing"))
print("Agent Decision for 'Red':", traffic_advice("Red"))
print("Agent Decision for 'Teal':", traffic_advice("Teal"))
```

**Step 9: Debugging the Knowledge Base** `[2 marks]`

```python
# You can test new or faulty signals to see how the agent reacts.
test_cases = ["Red", "Yellow", "Green", "Blue", "Teal", "None", "Purple"]
print("\n--- Debugging the Knowledge Base ---")
for color in test_cases:
    print(f"Light: {color:6} -> Action: {traffic_advice(color)}")
```

# Part 4: Uniform Cost Search — Nairobi to Kisumu

Find the cheapest route from Nairobi to Kisumu using Uniform Cost Search (distances in km).

```python
import heapq
#This dictionary represents the graph we made on the routes to Kisumu. The key is the town and the values represent all
#other towns that it can go to
graph = {
    "Nairobi": {"Thika": 45 , "Narok": 142 , "Naivasha": 90},
    "Thika": {"Nyeri": 105},
    "Nyeri": {"Nanyuki": 60, "Nyahururu": 100},
    "Nanyuki": {},
    "Nyahururu": {"Nakuru": 70},
    "Naivasha": {"Nakuru": 70},
    "Nakuru": {"Kericho": 107, "Eldoret": 156},
    "Narok": {"Kisii": 170},
    "Kericho": {"Kisii": 100, "Kisumu": 84},
    "Eldoret": {"Kakamega": 115, "Webuye": 60},
    "Webuye": {"Kakamega": 40},
    "Kisii": {"Kisumu": 112},
    "Kakamega":{"Kisumu": 52},
}
```

```python
#this is where we will store the candidate towns and decide which path to take(the cheapest)
priority_queue = []
#this ensures that we select the cheapest path in the heap
heapq.heappush(priority_queue, (0, "Nairobi", ["Nairobi"]))
#this is a dictionary that stores the cheapest path to a specific town from the start town(Nairobi)
best_cost = {
    "Nairobi": 0,
}

#This loops through all the available paths to Kisumu and picks the cheapest one
while priority_queue:
    # 1. Pop the cheapest entry
    current_cost,current_town,current_path = (heapq.heappop(priority_queue))
    # 2. Skip outdated entries
    if current_cost > best_cost[current_town]:
      continue
    # 3. Check if current town is Kisumu
    if current_town == "Kisumu":
      print(f"Minimum cost: {best_cost[current_town]}")
      print(f"Path: {current_path}")
      break
    # 4. Explore neighbouring towns
    for town, cost in graph[current_town].items():
      # 5. Calculate new costs
      new_cost = cost + current_cost
      if town not in best_cost or new_cost < best_cost[town]:
        # 6. Update best_cost and priority_queue
        best_cost[town] = new_cost
        new_path = current_path + [town]
        heapq.heappush(priority_queue, (new_cost,town,new_path))
```

# Knowledge engineering

## Smart traffic light navigation agent

This notebook covers steps 1–4 of the [source exercise](https://github.com/AllanVikiru/ArtificialIntelligence/blob/kba/Knowledge_Engineering.ipynb).

## Step 1. Identify the task

Given the observed color of a traffic light, the agent selects a driving action. The exercise models a single steady traffic signal with three possible colors. The goal is to make the agent's decision predictable for each of those colors.

## Step 2. Assemble relevant knowledge

The agent needs the observed color and the action associated with each color. In this exercise, red means stop, yellow means slow down and prepare to stop, and green means proceed if the intersection is clear. The model assumes that the observed signal applies to the agent's lane.

The rules come from the source exercise. The agent uses no APIs or web scraping to assemble its knowledge base.

## Step 3. Define the vocabulary

| Symbol | Meaning | Python representation |
| --- | --- | --- |
| `Red`, `Yellow`, `Green` | The three steady signal colors | String constants used as keys in `rules` |
| `Light(color)` | The agent observes a signal of the given color | An observed color would be stored as a string such as `current_light` |
| `CanGo(color)` | The color permits proceeding if the intersection is clear | The Boolean `can_go` stored in each rule |
| `Action(color)` | The action the agent should take for that color | The `action` text stored in each rule |

## Step 4. Encode general knowledge about the domain

The dictionary maps each color to its action and records whether that signal permits proceeding. Each rule applies when `Light(color)` matches the corresponding color. The output lists the encoded rules so they can be checked against Step 2.

```python
rules = {
    "Red": {"action": "Stop.", "can_go": False},
    "Yellow": {"action": "Slow down and prepare to stop.", "can_go": False},
    "Green": {"action": "Proceed if the intersection is clear.", "can_go": True},
}

for color, rule in rules.items():
    print(f"{color}: {rule['action']} CanGo({color}) = {rule['can_go']}")
```
